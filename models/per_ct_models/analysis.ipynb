```python
import pandas as pd
```

```python
csv_path = "/home/chb3333/yulab/chb3333/gem-patho/models/per_ct_models/global_test_cindices_by_cancer.csv"

# Load the DataFrame
df = pd.read_csv(csv_path)

# Drop NaNs from the c_index column
valid_cindices = df["c_index"].dropna()

# Compute mean and standard deviation
mean_ct_cindex = valid_cindices.mean()
std_ct_cindex = valid_cindices.std()

# Print the results
print(f"Mean per–cancer–type c-index: {mean_ct_cindex:.4f}")
print(f"STD  per–cancer–type c-index: {std_ct_cindex:.4f}")

# Save to file
output_path = "/home/chb3333/yulab/chb3333/gem-patho/models/per_ct_models/global_test_mean_std_cindex_by_cancer.txt"
with open(output_path, "w") as f:
    f.write(f"Mean per–cancer–type c-index: {mean_ct_cindex:.4f}\n")
    f.write(f"STD  per–cancer–type c-index: {std_ct_cindex:.4f}\n")

print(f"Saved mean and std of per–cancer–type c-indices to {output_path}")
```
