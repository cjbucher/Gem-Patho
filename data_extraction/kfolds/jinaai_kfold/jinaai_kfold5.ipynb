```python
gene_embs_path = "/home/chb3333/yulab/chb3333/gem-patho/data_extraction/genename_embeddings/text_embedding_models/jinaai/embeddings.parquet"
df_gene_embs = pd.read_parquet(gene_embs_path, engine="pyarrow")

# Convert the DataFrame to a dict: {gene_name -> np.ndarray}
gene_embeddings_dict = {
    gene: row.values
    for gene, row in df_gene_embs.iterrows()
}

print("Loaded gene embeddings for", len(gene_embeddings_dict), "genes")
```

```python
gene_synonyms = {
    "C15orf65": "CCDC186",
    "SEPT5": "SEPTIN5",
    "SEPT6": "SEPTIN6",
    "SEPT9": "SEPTIN9",
    "CARS": "CARS1"
}

# Genes that must be skipped because they cannot be translated
genes_to_skip = {
    "CTNNB1",
    "FGFR1OP",
    "H3F3A",
    "H3F3B",
    "HIST1H3B",
    "HIST1H4I",
    "HMGN2P46",
    "IGH",
    "IGK",
    "IGL",
    "TRA",
    "TRB",
    "TRD",
    "TRG",
    "H3P6",
    "NBEAP1",
    "NHERF1",
    "TERC",
    "TMSB4XP8"
}
```

```python

def resolve_gene_name(gene):
    """
    Given a gene symbol, return the translated symbol if available.
    Return None if the gene should be skipped.
    Otherwise, return the original gene symbol.
    """
    if gene in genes_to_skip:
        return None
    return gene_synonyms.get(gene, gene)

def map_polyphen_to_gene_embeds(polyphen_vector, gene_list, gene_embeddings):
    """
    Given:
      - polyphen_vector: an iterable of numeric scores, one per gene in 'gene_list'.
      - gene_list: the ordered list of genes matching positions in the polyphen vector.
      - gene_embeddings: dict {gene_name -> np.ndarray/list} with JinaAI gene embeddings.
    
    Returns:
      A list of dicts, each containing:
         { 'gene': <resolved_gene_name>,
           'embedding': <list_of_floats>,
           'score': <polyphen score> }
      ...for every gene with a nonzero polyphen score.
    """
    seq = []
    if len(polyphen_vector) != len(gene_list):
        print(f"Length mismatch: polyphen vector length {len(polyphen_vector)} != gene list length {len(gene_list)}")
        return seq

    for gene, score in zip(gene_list, polyphen_vector):
        if score == 0:
            continue

        resolved_name = resolve_gene_name(gene)
        if resolved_name is None:
            # Gene is in the skip set; skip it.
            continue

        if resolved_name in gene_embeddings:
            emb = gene_embeddings[resolved_name]
            if isinstance(emb, np.ndarray):
                emb = emb.tolist()
            seq.append({
                "gene": resolved_name,
                "embedding": emb,
                "score": score
            })
        else:
            print(f"Warning: Embedding for gene '{resolved_name}' not found.")
    return seq
```

```python
# Optionally check for missing genes in the gene list
missing_genes = [g for g in gene_list if resolve_gene_name(g) not in gene_embeddings_dict]
if missing_genes:
    print(f"There are {len(missing_genes)} genes missing embeddings after resolution. Some examples:")
    print(missing_genes[:10])
else:
    print("All genes from gene_list are present in the embeddings dictionary after resolution.")
```

```python
df_desc_subset = df_desc[["Case IDs", "CANCER_TYPE_ACRONYM", "description_embeddings"]].copy()

# Split the 'Case IDs' column (a comma-separated string) into a list and strip spaces
df_desc_subset["Case IDs"] = df_desc_subset["Case IDs"].apply(lambda x: [case.strip() for case in x.split(",")])

# Explode so that each Case ID is in its own row; then rename the column for merging
df_desc_exploded = df_desc_subset.explode("Case IDs").rename(columns={"Case IDs": "Case ID"})
print("Processed description embeddings shape:", df_desc_exploded.shape)

master_cols = ["polyphen_score", "mutated_genes", "OS.time", "OS", "type", "Case ID"]
df_master_subset = df_master[master_cols].copy()
```

```python
df_merged = pd.merge(df_test, df_desc_exploded, how="left", on="Case ID")
print("After merging test set with description embeddings:", df_merged.shape)

# Merge with the master mutation data (subset)
df_merged = pd.merge(df_merged, df_master_subset, how="left", on="Case ID")
print("Final merged data shape:", df_merged.shape)

```

```python
if "polyphen_score" in df_merged.columns:
    df_merged["gene_embed_seq"] = df_merged["polyphen_score"].apply(
        lambda vec: map_polyphen_to_gene_embeds(vec, gene_list, gene_embeddings_dict)
    )
    print("Gene embedding sequences have been mapped.")
else:
    print("Warning: 'polyphen_score' column not found in the merged DataFrame.")

# Show an example of the resulting data for one patient
print("Example gene embedding mapping for one patient:")
print(df_merged[["Case ID", "gene_embed_seq"]].head(1).to_dict(orient="records"))
```
