```python
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
base_path = "/home/chb3333/yulab/chb3333/gem-patho/data_extraction/kfolds/jinaai_kfold_20val"

# Get sorted list of fold directories
fold_dirs = sorted([d for d in os.listdir(base_path) if os.path.isdir(os.path.join(base_path, d))])
splits = ['train', 'val', 'test']

results_raw = {}
results_norm = {}
```

```python
dist_norm = {split: {} for split in splits}
dist_raw  = {split: {} for split in splits}

# Loop through each fold and each split to compute the distributions
for fold in fold_dirs:
    fold_path = os.path.join(base_path, fold)
    for split in splits:
        file_path = os.path.join(fold_path, f"{split}.parquet")
        df = pd.read_parquet(file_path)
        # Raw counts of each Project ID
        counts = df['Project ID'].value_counts().sort_index()
        dist_raw[split][fold] = counts
        # Normalized distribution (percentage)
        norm = (df['Project ID'].value_counts(normalize=True).sort_index() * 100).round(2)
        dist_norm[split][fold] = norm
```

```python
for split in splits:
    print(f"\n=== {split.upper()} SET DISTRIBUTION ACROSS FOLDS ===")
    # Combine raw counts and normalized percentages into DataFrames.
    df_raw = pd.DataFrame(dist_raw[split]).fillna(0).astype(int)
    df_norm = pd.DataFrame(dist_norm[split]).fillna(0)
    
    print("\nRaw Counts:")
    print(df_raw)
    print("\nNormalized Distribution (%):")
    print(df_norm)
    
    # Check for differences by computing basic summary statistics across folds
    # For normalized distribution, compute the mean and std deviation across folds for each Project ID
    stats = df_norm.copy()
    stats['Mean (%)'] = stats.mean(axis=1).round(2)
    stats['Std Dev (%)'] = stats.std(axis=1).round(2)
    
    print("\nSummary statistics across folds (Normalized %):")
    print(stats)
```
