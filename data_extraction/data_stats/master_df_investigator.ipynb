```python
results = {}

for cancer_type, group in df.groupby('type'):
    # Stack the numpy arrays in the group into a 2D array:
    mutated_matrix = np.stack(group['mutated_genes'].values)
    
    # Count the number of samples for this cancer type
    n_samples = mutated_matrix.shape[0]
    
    # Sum across samples (rows) for each gene (columns)
    gene_mut_counts = mutated_matrix.sum(axis=0)
    
    # Compute the percentage of samples with a mutation for each gene
    mutation_percentage = gene_mut_counts / n_samples * 100
    
    # Create a DataFrame with genes, their mutation counts, the sample count, and percentage
    gene_counts = pd.DataFrame({
        'gene': genes,
        'mutation_count': gene_mut_counts,
        'sample_count': n_samples,  # Same for all genes in this cancer type
        'mutation_percentage': mutation_percentage
    })
    
    # Sort the genes by mutation_count in descending order
    gene_counts = gene_counts.sort_values('mutation_count', ascending=False)
    
    # Save the result for the current cancer type
    results[cancer_type] = gene_counts
    
    csv_filename = os.path.join(output_dir, f"{cancer_type}_mutation_stats.csv")
    gene_counts.to_csv(csv_filename, index=False)
    print(f"Saved CSV for cancer type '{cancer_type}' to: {csv_filename}")
```

```python
all_mutations = np.stack(df['mutated_genes'].values)

# Sum across all samples (rows) for each gene (columns)
overall_gene_counts = all_mutations.sum(axis=0)

# Create a DataFrame mapping genes to their overall mutation counts
overall_counts_df = pd.DataFrame({
    'gene': genes,
    'overall_mutation_count': overall_gene_counts
})

# Count how many genes have an overall mutation count of 0
zero_mutation_genes_df = overall_counts_df[overall_counts_df['overall_mutation_count'] == 0]
num_zero_mutation_genes = zero_mutation_genes_df.shape[0]

print("Number of genes with a mutation count of 0 across all cancer types:", num_zero_mutation_genes)
print("\nGenes with 0 mutation count:")
print(zero_mutation_genes_df['gene'].tolist())
```

```python
cna_data_path = "/home/chb3333/yulab/chb3333/gem-patho/data_extraction/CNAs/cna_data.parquet"

cna_df = pd.read_parquet(cna_data_path)
```

```python
cna_df
```

```python
cancer_type_mapping = df.set_index('Case ID')['type']
cna_df['type'] = cna_df['normalized_patient_id'].map(cancer_type_mapping)

results = {}

for cancer_type, group in df.groupby('type'):
    # --- Mutation Statistics (from df) ---
    mutated_matrix = np.stack(group['mutated_genes'].values)
    n_samples_mut = mutated_matrix.shape[0]
    gene_mut_counts = mutated_matrix.sum(axis=0)
    mutation_percentage = gene_mut_counts / n_samples_mut * 100
    
    gene_counts = pd.DataFrame({
        'gene': genes,
        'mutation_count': gene_mut_counts,
        'sample_count': n_samples_mut,
        'mutation_percentage': mutation_percentage
    })
    gene_counts = gene_counts.sort_values('mutation_count', ascending=False)
    
    # --- CNA Statistics (from cna_df) ---
    # Filter CNA data for the current cancer type
    cna_group = cna_df[cna_df['type'] == cancer_type]
    n_samples_cna = cna_group.shape[0]
    
    # For each gene in our mutation gene list, count non-zero CNA events
    cna_counts = []
    cna_percentages = []
    for gene in genes:
        # Check if the gene exists as a column in the CNA dataframe
        if gene in cna_group.columns:
            # Count samples with a non-zero CNA value for this gene
            count = (cna_group[gene] != 0).sum()
            
        else:
            count = 0
        cna_counts.append(count)
        # Compute percentage if there are any CNA samples for this type
        perc = (count / n_samples_cna * 100) if n_samples_cna > 0 else 0
        cna_percentages.append(perc)
    
    cna_stats_df = pd.DataFrame({
        'gene': genes,
        'cna_count': cna_counts,
        'cna_percentage': cna_percentages
    })
    
    # --- Combine the Mutation and CNA Stats ---
    combined_stats = pd.merge(gene_counts, cna_stats_df, on='gene', how='left')
    combined_stats[['cna_count', 'cna_percentage']] = combined_stats[['cna_count', 'cna_percentage']].fillna(0)
    
    # Save the result for the current cancer type
    results[cancer_type] = combined_stats
    
    # Save the combined DataFrame as CSV
    csv_filename = os.path.join(output_dir, f"{cancer_type}_mutation_cna_stats.csv")
    combined_stats.to_csv(csv_filename, index=False)
    print(f"Saved CSV for cancer type '{cancer_type}' to: {csv_filename}")
```

```python
# %% [markdown]
# ### Calculate Patient-Level Mutation Counts for Each Gene
# 
# Here we convert each patient’s mutation vector (stored in 'mutated_genes') to binary (1 if mutated, 0 otherwise)
# and sum across patients to get the number of patients with at least one mutation per gene.

# %%
# Ensure you're using the original DataFrame that still has 'mutated_genes'
df['binary_mutated_genes'] = df['mutated_genes'].apply(lambda arr: (arr > 0).astype(int))

# %%
# Stack the binary arrays and sum along the patient axis to obtain patient counts per gene
all_binary_mutations = np.stack(df['binary_mutated_genes'].values)
patient_mutation_counts = all_binary_mutations.sum(axis=0)

# Create a DataFrame mapping genes to the number of patients with a mutation
genes = gene_df['Gene Symbol'].tolist()  # Assumes gene_df was loaded earlier

patient_mutation_df = pd.DataFrame({
    'gene': genes,
    'patient_mutation_count': patient_mutation_counts
})

# Display the top 10 genes by patient mutation count
print(patient_mutation_df.sort_values("patient_mutation_count", ascending=False).head(10))
```

```python
# %% [markdown]
# ### Calculate Patient-Level CNA Counts for Each Gene
# 
# For CNA data we similarly convert the values to binary (1 if there is any CNA event, 0 otherwise)
# and count the number of patients with a CNA event for each gene.

# %%
# Ensure the gene list is defined
genes = gene_df['Gene Symbol'].tolist()

# %%
# Count the number of patients with a non-zero CNA value for each gene
patient_cna_counts = {}
for gene in genes:
    if gene in cna_df.columns:
        patient_cna_counts[gene] = (cna_df[gene] != 0).sum()
    else:
        patient_cna_counts[gene] = 0

patient_cna_df = pd.DataFrame({
    'gene': list(patient_cna_counts.keys()),
    'patient_cna_count': list(patient_cna_counts.values())
})

# Display the top 10 genes by patient CNA count
print(patient_cna_df.sort_values("patient_cna_count", ascending=False).head(10))
```
